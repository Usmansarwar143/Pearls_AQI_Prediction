# Exploratory Data Analysis (EDA) - Pearls AQI Predictor
Let's load the data using our pipeline scripts and visualize some correlations.

```python
import sys
sys.path.append('../src/feature_pipeline')

from fetch_api import get_data
from compute_features import engineer_features

# Fetch 30 days of data for EDA
p_df, w_df = get_data(days_back=30)
df = engineer_features(p_df, w_df)
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='date', y='aqi')
plt.title('AQI over Time')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
cols_to_plot = ['aqi', 'temperature_2m', 'relative_humidity_2m', 'wind_speed_10m', 'co', 'pm2_5', 'pm10']
sns.heatmap(df[cols_to_plot].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
